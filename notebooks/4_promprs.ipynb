```python
from dotenv import load_dotenv
load_dotenv()

from langchain_openai import ChatOpenAI

llm = ChatOpenAI(model="gpt-5.4-mini", stream_usage=True,temperature=0.7)
```

```python
from langchain_core.prompts import ChatPromptTemplate

my_prompt = ChatPromptTemplate(
    [
        ("system", "You are a helpful AI bot. that helps to create a curated post for a social media platform like linkedIn"),
        ("human", "I want to cretae a post about {topic}"),
    ]
)
```

```python
my_prompt.invoke({"topic":"data engineering"})
```

```python
chain=my_prompt | llm
```

```python
chain.invoke({"topic":"The importance of data privacy in digital age"})
```
