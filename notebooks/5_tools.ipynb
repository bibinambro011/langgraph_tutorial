```python
from dotenv import load_dotenv
load_dotenv()

from langchain_openai import ChatOpenAI

llm = ChatOpenAI(model="gpt-5.4-mini", stream_usage=True,temperature=0.7)
```

```python
from langchain_tavily import TavilySearch

```

```python
TavilySearch().invoke("tell me about todays wether in ernakulam")
```

```python
from langchain_community.tools import tool

@tool
def tavily_function(query:str) -> str:
    """"use this tool to search the web for the latest news and information"""
    return TavilySearch().invoke(query)
```

```python
tools=[tavily_function]
```

```python
bind_tools=llm.bind_tools(tools)

bind_tools.invoke("what is teh current weather in ernalulam")
```
